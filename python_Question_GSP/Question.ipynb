# 13. Reverse a Number

```python
num = int(input("Enter a number: "))

reverse = 0

while num > 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num = num // 10

print("Reversed Number:", reverse)
```

# 14. Palindrome cheacker

```python
string = input("Enter a string: ")

if string == string[::-1]:
    print("Palindrome")
else:
    print("Not Palindrome")

# without slicing
string = input("Enter a string: ")

reverse = ""

for ch in string:
    reverse = ch + reverse

if string == reverse:
    print("Palindrome")
else:
    print("Not Palindrome")
```

# 15 find max,min,average

```python
numbers = [10, 20, 5, 40, 15]

largest = numbers[0]
smallest = numbers[0]
total = 0

for i in numbers:
    if i > largest:
        largest = i

    if i < smallest:
        smallest = i

    total += i

average = total / len(numbers)

print("Largest element:", largest)
print("Smallest element:", smallest)
print("Average:", average)
```

# 19. count vowels and consonent in a string

```python
string = input("Enter a string: ")

vowels = 0
consonants = 0

for ch in string.lower():
    if ch.isalpha():          # Check only letters
        if ch in "aeiou":
            vowels += 1
        else:
            consonants += 1

print("Number of vowels:", vowels)
print("Number of consonants:", consonants)
```

# 20. remove duplicate from list

```python
numbers = [1, 2, 3, 2, 4, 1, 5, 3]

unique = []

for i in numbers:
    if i not in unique:
        unique.append(i)

print("Original List:", numbers)
print("List after removing duplicates:", unique)
```

# 21. sort a list without using build in sort function

```python
numbers = [5, 2, 8, 1, 9, 3]

print("Before Sorting:", numbers)

n = len(numbers)

for i in range(n):
    for j in range(n - i - 1):
        if numbers[j] > numbers[j + 1]:
            # Swap
            temp = numbers[j]
            numbers[j] = numbers[j + 1]
            numbers[j + 1] = temp

print("After Sorting:", numbers)
```
